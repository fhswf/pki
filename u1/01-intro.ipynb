**Übung 7: Was wird das folgende Programm ausgeben:**

```python
x = 43
x = x + 1
print(x)
```

a\) 43\
b) 44\
c) x + 1\
d) Fehler, denn x = x + 1 ist mathematisch nicht möglich

<details>
<summary><b>Antwort anzeigen</b></summary>

**Antwort: b) 44**

Die zweite Zuweisung erhöht den bisherigen Wert von `x` um 1. Deshalb gibt `print(x)` den Wert 44 aus.

</details>
