### Desktop: tkinter

````python
%%writefile gui_app_tkinter.py
import tkinter as tk
from tkinter import ttk

def save_callback():
    print("Save clicked:", entry.get())

def slided(value):
    scale_label.config(text=f"Wert: {float(value):.2f}")

root = tk.Tk()
root.title("Tkinter Beispiel")
entry = ttk.Entry(root)
scale_label = ttk.Label(root, text="Wert:")
scale = ttk.Scale(root, from_=0, to=100, command=slided)
button = ttk.Button(root, text="Save", command=save_callback)
for widget in (entry, scale_label, scale, button):
    widget.pack(pady=5)
root.mainloop()
````

### Web: Streamlit

````python
%%writefile gui_app_streamlit.py
import streamlit as st

st.title("Streamlit Beispiel")
name = st.text_input("Name", "Welt")
wert = st.slider("Wert", 0, 100, 42)
if st.button("Save"):
    st.success(f"Gespeichert: {name}, {wert}")
st.write(f"Hallo {name}! Der Wert ist {wert}.")
````

## Heute im Praktikum: Regex

- Regex sind mächtige Werkzeuge zur Textsuche und -verarbeitung
- Viele klassische Regex-Aufgaben lassen sich heute direkt mit LLMs lösen
- LLMs erkennen Muster, extrahieren Informationen und transformieren Texte ohne komplexe Syntax
- Regex bleiben wichtig, wenn Präzision, Geschwindigkeit oder feste Regeln gefordert sind
- LLMs benötigen entweder Internetzugang zur API oder lokale Rechenressourcen

Zugang zum AI-Hub über Umgebungsvariablen (z. B. aus einer Datei `.env`):

```
AI_HUB_API_KEY=<eigener Schlüssel>
AI_HUB_BASE_URL=https://hub.ki.fh-swf.de/v1
AI_HUB_MODEL=<kleines Modell>
```

````python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
client = OpenAI(api_key=os.environ["AI_HUB_API_KEY"], base_url=os.environ["AI_HUB_BASE_URL"])
modell = os.environ["AI_HUB_MODEL"]

code = "for i in range(0,10): print(i)"

resp = client.chat.completions.create(
    model=modell,
    messages=[{"role": "user", "content": f"Verbessere diesen Code:\n{code}"}]
)
print(resp.choices[0].message.content)
````

## SSH-Schlüssel einrichten

Anmeldung im Browser per **SSO** – Git authentifiziert sich per **SSH-Schlüssel**.
Privater Schlüssel bleibt lokal, öffentlicher kommt ins Profil.

````python
import os, subprocess
from pathlib import Path

PLATTFORM = "gitlab"                      # "gitlab" oder "github"
HOST = {"gitlab": "gitlab.git.nrw", "github": "github.com"}[PLATTFORM]
NAME  = "Heiner Giefers"
EMAIL = "giefers.heiner@fh-swf.de"

os.environ.update(HOST=HOST, NAME=NAME, EMAIL=EMAIL)   # für die Shell-Zellen 
````

Bei GitHub kann man als E-Mail die noreply-Adresse aus den E-Mail-Einstellungen verwenden.

````python
schluessel = Path.home() / ".ssh" / "gitlab-test-key"

if schluessel.exists():
    print("Schlüssel existiert bereits – wird weiterverwendet.")
else:
    schluessel.parent.mkdir(mode=0o700, exist_ok=True)
    subprocess.run(["ssh-keygen", "-t", "ed25519", "-C", EMAIL,
                    "-f", str(schluessel), "-N", ""], check=True)

print(schluessel.with_suffix(".pub").read_text())
````

**Im Browser:** Profil → **SSH-Schlüssel** → öffentlichen Schlüssel einfügen.

Ohne Passphrase, weil das Notebook keine Passworteingabe weiterreicht. Im eigenen Terminal besser mit Passphrase und ssh-agent bzw. macOS-Schlüsselbund.
Ein vorhandener Schlüssel mit Passphrase funktioniert im Notebook nur, wenn er im ssh-agent geladen ist.

````python
# Eigener Kurzname in ~/.ssh/config, damit Git genau diesen Schlüssel verwendet
SSH_HOST = f"{PLATTFORM}-test"
config = Path.home() / ".ssh" / "config"
eintrag = (f"\nHost {SSH_HOST}\n"
           f"  HostName {HOST}\n"
           f"  User git\n"
           f"  IdentityFile {schluessel}\n"
           f"  IdentitiesOnly yes\n"
           f"  StrictHostKeyChecking accept-new\n")

if config.exists() and SSH_HOST in config.read_text().split():
    print(f"Eintrag „{SSH_HOST}“ existiert bereits.")
else:
    with config.open("a") as f:
        f.write(eintrag)
    config.chmod(0o600)
    print(eintrag)
os.environ["SSH_HOST"] = SSH_HOST
````

Der Kurzname (z. B. `gitlab-test`) zeigt auf denselben Server, nutzt aber nur den Schlüssel aus dem Notebook (`IdentitiesOnly yes`). Andere Einträge in `~/.ssh/config`, etwa für `gitlab.git.nrw` selbst, bleiben unberührt.
`StrictHostKeyChecking accept-new` speichert den Server-Fingerabdruck bei der ersten Verbindung ohne Rückfrage.
In SSH-Adressen aus dem Browser den Servernamen durch den Kurznamen ersetzen: `git@gitlab-test:…` statt `git@gitlab.git.nrw:…`.

````python
%%bash
ssh -T $SSH_HOST
````

Erwartet: „Welcome to GitLab, @…“ bzw. bei GitHub „Hi …! You've successfully authenticated“.

## Demo 1: klonen, ändern, pushen

**Im Browser:** leeres Projekt anlegen (privat, mit README), SSH-Adresse kopieren.

````python
GIT_USERNAME = "giefers.heiner" # nur für gitlab.git.nrw
URL  = f"git@{SSH_HOST}:{GIT_USERNAME}/pki-test-projekt.git"   # <- Pfad aus der SSH-Adresse im Browser
REPO = "gitlab-demo"
os.environ.update(URL=URL, REPO=REPO)
README = Path(REPO) / "README.md"

def anhaengen(text):
    """Text an die README anhängen – nur, falls er noch fehlt (Zelle wiederholbar)."""
    if text not in README.read_text(encoding="utf-8"):
        with README.open("a", encoding="utf-8") as f:
            f.write(text)
````

````python
%%bash
[ -d $REPO ] || git clone $URL $REPO
cd $REPO
git config user.name "$NAME"
git config user.email "$EMAIL"
git log --oneline
````

````python
anhaengen("\n## Notizen\n\nErster Commit aus dem Notebook.\n")
````

````python
%%bash
cd $REPO
git status --short
git diff
````

````python
%%bash
cd $REPO
git add README.md
git commit -m "README um Notizen ergänzt"
git push
````

- `add` = vormerken
- `commit` = lokal speichern
- `push` = auf den Server übertragen

## Demo 2: Issue → Branch → Merge Request

**Im Browser:** Issue anlegen, z. B. „Git-Befehle in die README“ – es bekommt die Nummer **#1**.

````python
%%bash
cd $REPO
git switch -C 1-git-befehle        # Issue-Nummer vorn im Namen
````

````python
anhaengen("\n## Git-Befehle\n\n`clone` · `add` · `commit` · `push` · `pull`\n")
````

````python
%%bash
cd $REPO
git commit -am "Git-Befehle ergänzt"
git push -u origin 1-git-befehle
````

`switch -C` statt `-c`, damit die Zelle wiederholbar ist. Normalerweise: `git switch -c <name>`.
Die Ausgabe von `git push` enthält den Link zum Anlegen des Merge Requests (GitHub: Pull Request).

**Im Browser:** Link aus der Ausgabe öffnen → Merge Request mit **`Closes #1`** → Änderungen ansehen, kommentieren → **Merge** → Issue ist geschlossen.

````python
%%bash
cd $REPO
git switch main
git pull
git log --oneline --graph
````

## Ausblick: CI/CD

Eine Datei im Repository – die Plattform testet jeden Push automatisch.

**GitLab:** `.gitlab-ci.yml`
```yaml
test:
  image: python:3
  script:
    - pip install pytest
    - pytest
```

**GitHub:** `.github/workflows/test.yml`
```yaml
on: push
jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v5
      - uses: actions/setup-python@v6
        with:
          python-version: "3.x"
      - run: pip install pytest
      - run: pytest
```

Ergebnis erscheint als Pipeline bzw. Check im Merge Request; typische Regel: Merge nur bei Grün.
Voraussetzung: Tests im Repository (`test_*.py`). Ohne Tests endet pytest mit Fehler und die Pipeline wird rot.
Die Versionsnummern der Actions ändern sich gelegentlich; die aktuelle steht in der Beschreibung der Action auf GitHub.

## Empfehlung: gitlab.git.nrw für Ihr Projekt

- Anmeldung mit dem Hochschulkonto per **SSO** – kein zusätzliches Konto nötig
- Projekt **privat** anlegen und die Teammitglieder einladen

**Im Team arbeiten**

- Aufgaben als **Issues** festhalten und verteilen
- Änderungen über **Branches** und **Merge Requests** zusammenführen
- Gegenseitig Code ansehen und kommentieren

**Abgabe weiterhin als ZIP-Datei in Moodle**

```bash
git archive --format=zip -o abgabe.zip HEAD
```

Teammitglieder lassen sich in GitLab meist erst einladen, wenn sie sich einmal angemeldet haben.
`git archive` packt nur, was eingecheckt ist – `.venv`, `.env` und andere ignorierte Dateien bleiben automatisch draußen.
