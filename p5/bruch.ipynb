## Bruch-Klasse

In diesem Arbeitsblatt soll eine Klasse für mathematische Brüche erstellt werden.
Ein Bruch wie $\frac{3}{4}$ besteht aus einem Zähler (hier 3) und einem Nenner (hier 4).
Die Klasse `Bruch` besitzt ein Klassen-Attribut `klassenname` und soll folgende Instanz-Attribute erhalten:
- `zaehler`: Der Zähler des Bruchs
- `nenner`: Der Nenner des Bruchs

In der folgenden Code-Zelle sind ebenfalls die Methoden `__str__` und `__repr__` implementiert.
Dabei handelt es um sogenannte *Magic Methods*, das sind Methoden mit einem festem Namen und fester Bedeutung.
Damit sie sich direkt von anderen Methoden abheben, beginnen und enden die Bezeichner von *Magic Methods* immer mit zwei Unterstrichen.
*Magic Methods* werden zu bestimmten Ereignissen aufgerufen.
Wir können diese *Magic Methods* (und *Magic Attributes*) überschreiben, um die vordefinierte Funktionalität zu ersetzen oder zu erweitern.

Die `__init__(self,...)` Funktion ist ein Beispiel für eine *Magic Method*. Wenn Sie eine neue Instanz einer Klasse erzeugen, sucht der Interpreter nach einer Methode mit genau diesem Namen und führt sie aus.
Die *Magic Methods* `__str__` und `__repr__` werden aufgerufen, wenn ein Objekt zu einem String umgewandelt wird, bzw. wenn die Darstellung des Objekts auf der Kommandozeile ausgegeben wird.

#### Aufgabe: Konstruktor vervollständigen
Programmieren Sie den Konstruktor der Klasse `Bruch` aus.
Der Konstruktor wird mit zwei Argumenten aufgerufen, mit denen der Zähler und der Nenner gesetzt werden. Das Bruch-Objekt soll diese Werte in den Attributen `zaehler` bzw. `nenner` abspeichern.
Der Bruch soll dabei **immer gekürzt** gespeichert werden. Dazu teilen Sie Zähler und Nenner durch ihren größten gemeinsamen Teiler (ggT).
Den ggT berechnet die Funktion `gcd` (*greatest common divisor*) aus dem Modul `math`:

```python
>>> gcd(6, 8)

```


##### Hinweis:

Der Konstruktor soll auch funktionieren, wenn kein Nenner angegeben wird. In diesem Fall wird der Nenner 1 verwendet, der Bruch entspricht dann einer ganzen Zahl.
Außerdem darf der Nenner niemals 0 sein und ein negatives Vorzeichen soll immer im Zähler stehen. Dieses Verhalten ist in der folgenden Code-Zelle bereits implementiert.

```python
b = Bruch(5)
```

*Erwartete Ausgabe:*


````python
from math import gcd


class Bruch:

    klassenname = "Bruch"

    # Hier soll der Konstruktor der Klasse angegeben werden, also
    # def __init__(...)
    def __init__(self, zaehler, nenner=1):
        if not isinstance(zaehler, int) or not isinstance(nenner, int):
            raise TypeError("Zähler und Nenner müssen ganze Zahlen sein")
        if nenner == 0:
            raise ZeroDivisionError("Der Nenner darf nicht 0 sein")
        if nenner < 0:
            zaehler = -zaehler
            nenner = -nenner
        # YOUR CODE HERE
        raise NotImplementedError()

    # __str__() wird aufgerufen, wenn ein Objekt in eine Zeichenkette
    # umgewandelt werden muss. Dies ist z.B. bei 'print' der Fall
    def __str__(self):
        if self.nenner == 1:
            return f"{self.zaehler}"
        return f"{self.zaehler}/{self.nenner}"

    # __repr__() soll eine möglichst eindeutige Repräsentation des Objekts
    # als Zeichenkette liefern. Diese Ausgabe erscheint auf der Kommandozeile
    # als Ausgabe für das Objekt
    def __repr__(self):
        return f"{self.klassenname}-Objekt ({id(self)}): {self.__str__()}"
````

````python
# Test Konstruktor
b0 = Bruch(6, 8)
assert b0.zaehler==3 and b0.nenner==4, "Der Bruch wurde nicht gekürzt"
b1 = Bruch(-10, 4)
assert b1.zaehler==-5 and b1.nenner==2
b2 = Bruch(3, -9)
assert b2.zaehler==-1 and b2.nenner==3, "Das Vorzeichen muss im Zähler stehen"
print(b0, b1, b2)
````

#### Eigene Methode definieren

Für spätere Aufgaben ist es hilfreich, wenn wir den Wert eines Bruchs als Dezimalzahl (`float`) erhalten können.
Daher fügen wir der Klasse `Bruch` eine neue Methode `als_float` hinzu, die den Zähler durch den Nenner teilt und das Ergebnis zurückliefert.

Mit dem Decorator `@property` können wir eine Methode wie ein Attribut verwenden: Die Property `F` ruft `als_float()` auf und liefert den Wert als Gleitkommazahl. Statt `b.als_float()` können wir dann `b.F` schreiben – ohne Klammern.

**Hinweis:** Damit wir die Klasse `Bruch` in mehreren Code-Zellen entwickeln können, müssen wir die ursprüngliche Klasse `Bruch` erweitern.
Dazu erzeugen wir jeweils eine neue Klasse `Bruch` und *erben* alles aus der zuvor definierten Klasse `Bruch`. Das geht, indem wir bei der Klassen-Definition die Vater-Klasse `Bruch` angeben:
```python
class Bruch(Bruch):
    pass
```

````python
class Bruch(Bruch):

    def als_float(self):
        return self.zaehler / self.nenner

    @property
    def F(self):
        return self.als_float()
````

````python
# Test als_float
b0 = Bruch(3, 4)
res = b0.als_float()
assert res==0.75
assert b0.F==0.75
print(res)
````

#### Operatoren überladen

Die *Magic Methods* in Python haben noch eine weitere Funktion, mit ihnen können die bekannten Operatoren **überladen** werden.
Das bedeutet, wir können z.B. selbst definieren, was der *Plus-Operator* `+` für unsere Klasse für eine Bedeutung hat.
Wir müssen dafür einfach die Methode `__add__` definieren.
`__add__` muss neben dem `Bruch`-Objekt `self` noch ein weiteres Objekt erhalten und unbedingt ein neues Objekt zurückliefern.
Das passt genau zu unserer `add`-Methode.
Wir können also einfach `add` in `__add__` aufrufen:

````python
class Bruch(Bruch):

    def __add__(self, other):
        return self.add(other)
````

Nun überprüfen wir, ob das Ergebnis passt. Zum Vergleich rechnen wir dieselbe Aufgabe auch mit Gleitkommazahlen:

````python
print("Gleitkommazahlen:", 0.1 + 0.2, "  gleich 0.3?", 0.1 + 0.2 == 0.3)
print("Brüche:          ", Bruch(1, 10) + Bruch(2, 10))
````

Gleitkommazahlen werden im Rechner im Binärsystem gespeichert. Dezimalzahlen wie 0.1 lassen sich dort nicht exakt darstellen, daher entstehen kleine Rundungsfehler.
Brüche aus ganzen Zahlen rechnen dagegen **exakt**. Deshalb lassen wir in unserer Klasse auch keine `float`-Werte zu.

Für unsere `Bruch`-Klasse können wir noch viele weitere *Magic Methods* ausimplementieren.
Z.B. die Methoden `__eq__`, um Brüche auf Gleichheit zu überprüfen oder `__lt__` (*less than*), die beim Vergleich mit `<` aufgerufen wird.
Ebenfalls hilfreich ist die Methode `__float__`, die aufgerufen wird, wenn ein Objekt mit `float(...)` in eine Gleitkommazahl umgewandelt wird.

Die Methoden `__radd__` und `__rmul__` werden aufgerufen, wenn unser Bruch auf der **rechten** Seite des Operators steht, z.B. bei `2 * Bruch(1, 3)`. Python fragt dann zunächst die Zahl 2, ob sie mit einem Bruch multipliziert werden kann. Weil `int` unsere Klasse nicht kennt, wird anschließend `__rmul__` des Bruchs aufgerufen.

Tipp: Weil unsere Brüche immer gekürzt gespeichert werden, sind zwei Brüche genau dann gleich, wenn Zähler und Nenner übereinstimmen.

````python
class Bruch(Bruch):

    def __lt__(self, other):
        if isinstance(other, int):
            other = Bruch(other)
        return self.zaehler*other.nenner < other.zaehler*self.nenner

    def __float__(self):
        return self.als_float()

    def __radd__(self, other):
        return self.add(other)

    def __rmul__(self, other):
        return self.mul(other)

    def __eq__(self, other):
        if isinstance(other, int):
            # Hier weiß man, dass 'other' eine ganze Zahl ist. Mache also aus 'other' einen Bruch. Dann kann man die beiden Brüche vergleichen.
            other = Bruch(other)
        if isinstance(other, self.__class__):
            # Hier weiß man, dass 'other' ein Bruch ist. Die Gleichheit zweier Brüche erfolgt, wenn ihre Zähler und Nenner übereinstimmen.
            # YOUR CODE HERE
            raise NotImplementedError()
        else:
            raise TypeError
    
````

#### Die Bruch-Klasse verwenden

Durch die *Magic Methods* verhalten sich unsere Brüche jetzt fast wie die eingebauten Zahlentypen von Python.
Viele Funktionen von Python funktionieren daher ohne weiteren Aufwand auch mit unserer Klasse:
- `sorted` sortiert eine Liste von Brüchen, weil es `__lt__` verwendet.
- `sum` addiert eine Liste von Brüchen. `sum` beginnt dabei mit der Zahl 0, also wird zuerst `0 + Bruch(...)` berechnet. Dafür benötigen wir `__radd__`.

````python
brueche = [Bruch(3, 4), Bruch(1, 3), Bruch(5, 6), Bruch(1, 2)]
print("Sortiert:", *sorted(brueche))
print("Summe:   ", sum(brueche))
````

Zum Abschluss ein kleines Experiment: Wir berechnen eine Näherung für $\pi$ mit der **Leibniz-Reihe**:

$\pi = 4 \cdot \sum_{k=0}^{\infty} \frac{(-1)^k}{2k+1} = 4 \cdot \left(1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \dots\right)$

Da jeder Summand ein Bruch ist, können wir unsere `Bruch`-Klasse verwenden. Die Reihe konvergiert allerdings sehr langsam. Wir geben die Näherung nach 10.000 Summanden als Gleitkommazahl aus.

````python
pi = Bruch(0)
for k in range(10000):
    summand = Bruch((-1)**k, 2*k + 1)
    pi = pi + 4 * summand

print(f"pi ≈ {float(pi):.15f}")
````
